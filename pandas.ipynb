```python
employees = pd.DataFrame({
 'employee_id':[1,2,3,4,5],
 'name':['john', 'Anna', 'Peter', 'Linda', 'Bob'],
 'department':['HR', 'IT', 'Finanace', 'IT', 'HR']})
salary = pd.DataFrame({
    'employee_id':[1,2,3,6,7],
    'salary':[60000,80000,65000,70000,90000],
    'bonus':[5000,10000,5000,10000,5000]
})
```

```python
employees
```

```python
salary
```

```python
# merge the common rows
# on must be common

# how refer to the type of merging means inner mergin means only the common part are show while the outer means all common non common parts are shown
# inner means like AND and outer is OR
```

```python
pd.merge(employees,salary, on = 'employee_id',how= 'outer')
```

```python
pd.merge(employees,salary, on = 'employee_id',how= 'left')
# how = 'left'means that the left table  rows are show but the value are according to the common or how = 'right'
```

```python
pd.merge(employees,salary, on = 'employee_id',how= 'inner')
```

```python
#  concatination
```

```python
pd.concat([employees , salary])
# on the basis of rows
```

```python
pd.concat([employees , salary],axis= 1)
# on the basis of rows
```
