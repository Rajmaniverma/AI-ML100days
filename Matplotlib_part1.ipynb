```python
# sub plot
```

```python
import matplotlib.pyplot as plt

plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [10, 20, 30])

plt.subplot(2, 2, 2)
plt.bar(["A", "B", "C"], [10, 20, 15])

plt.subplot(2, 2, 3)
plt.scatter([1, 2, 3], [5, 15, 10])

plt.subplot(2, 2, 4)
plt.hist([1, 2, 2, 3, 3, 3])

plt.show()
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1, 0.8 , 0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4, 0.2, 0.2])
axis2.plot(x,y)
```

```python
months = [1, 2, 3, 4, 5]

sales = [100, 120, 150, 180, 200]
profit = [20, 30, 40, 50, 60]

plt.plot(months, sales)
plt.plot(months, profit)

plt.show()
```

```python
plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend()

plt.show()
```
