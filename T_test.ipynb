```python
import numpy as np
```

```python
from scipy import stats
```

```python
sample = [ 172,174,168,169,171,173,175,170,169,172]
```

```python
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof =1 )
n= len(sample)
mean_population = 170
```

```python
sample_mean
```

```python
sample_std
```

```python
n
```

```python
t =((sample_mean - mean_population)*np.sqrt(n))/sample_std
```

```python
t
```

```python
P_value= 2*(1-stats.t.cdf(abs(t),df = n-1))
```

```python
P_value
alpha = 0.05
```

```python
if P_value < alpha :
  print("reject null hypothesis")
else:
  print("accept null hypothesis")
```
