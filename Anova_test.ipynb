```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df_clean = df[['age', 'pclass']].dropna()

df_clean
```

```python
df['pclass'].unique()
```

```python
class1 = df[df['pclass']==1 ]['age']
class2 = df[df['pclass']== 2]['age']
class3 = df[df['pclass']== 3]['age']
```

```python
class1 = df[df['pclass'] == 1]['age'].dropna()

class2 = df[df['pclass'] == 2]['age'].dropna()

class3 = df[df['pclass'] == 3]['age'].dropna()
```

```python
class2
```

```python
f_stats , p_value = f_oneway(class1,class2,class3)
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
  print(" reject the null hypothesis  and there is a signafacant difference in atleast " )
else:
  print("thereis no signifact difference ")
```
