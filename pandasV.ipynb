```python
df = pd.DataFrame({
    "department": ["CS", "CS", "IT", "IT", "CS"],
    "name": ["A", "B", "C", "D", "E"],
    "marks": [80, 90, 70, 85, 75]
})
```

```python
df.groupby("department")["marks"].mean()
```

```python
df.groupby("department")["marks"].sum()
```

```python
df.groupby("department")["marks"].agg(
    ["mean", "sum", "min", "max"]
)
```

```python
df1 = pd.DataFrame({
    "name": ["A", "B"],
    "marks": [80, 90]
})

df2 = pd.DataFrame({
    "name": ["C", "D"],
    "marks": [70, 85]
})

result = pd.concat([df1, df2])

print(result)
```

```python
students = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["A", "B", "C"]
})

marks = pd.DataFrame({
    "id": [1, 2, 3],
    "marks": [80, 90, 75]
})

result = pd.merge(students, marks, on="id")

print(result)
```

```python
def grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    else:
        return "C"

df["grade"] = df["marks"].apply(grade)
```
